Problem 1
Create a 5x5 array filled with random integers between 1 and 100. Then:
1. Find the maximum value and its position (row, column)
2. Replace all values greater than 50 with 0
3. Calculate the sum of each row

```python
import numpy as np
a = np.random.randint(1,100,size = (5,5))
print(a)
```

###1. Find the maximum value and its position (row, column)

```python
print("Max:", a.max(), "Position :",np.unravel_index(a.argmax(), a.shape))
print("Max Per Row : ", a.max(axis = 1), "Positions : ", a.argmax(axis = 1))
print("Max Per Column : ", a.max(axis = 0), "Positions : ", a.argmax(axis = 0))
```

```python
b = a.copy()
a[a > 50] = 0
print(a)
```

```python
print("Sum of each row : ", a.sum(axis = 1))
```

```python
flat_idx = a.argmax()
position = np.unravel_index(flat_idx, a.shape)
print(flat_idx)
print(position)
```

##Problem 2:
Create two 1D arrays of size 5 with random integers between 1-10. Then:

Calculate their dot product
Stack them vertically to form a 2x5 array
Transpose that array to get 5x2
Find the sum of each row in the transposed array

```python
array1 = np.random.randint(1,10,size=(5))
array2 = np.random.randint(1,10,size=(5))
print(array1)
print(array2)
```

```python
print("Dot Product : ",np.dot(array1,array2))
```

```python
array = np.stack([array1,array2])
print("Stack :",array)
```

```python
array = array.T
print( array )
```

```python
print(np.sum(array,axis = 1))
print(array.sum(axis = 1))
```

Problem 3:
You have a chess board represented as an 8x8 array of zeros. Place:

Your pieces (value 1) on row 0 and row 1 (all squares)
Opponent pieces (value -1) on row 6 and row 7 (all squares)

Then:

Print the board
Use boolean indexing to count how many squares belong to you (value 1)
Use np.where to create a new array where your pieces become 100, opponent pieces become -100, empty stays 0

```python
board = np.zeros([8,8])
print(board)
```

```python
board[:2,:] = 1
board[6:,:] = -1
print(board)
```

```python
print((board[board == 1]).shape[0])
```

```python
board = np.where(board > 0,100,np.where(board == 0, 0, -100))
print(board)
```

Problem 4:
Create a 3x3 identity matrix (you've seen this concept — matrix with 1s on diagonal, 0s elsewhere). Then:

Verify that multiplying any 3x3 matrix by the identity matrix gives back the same matrix
Calculate the determinant of the identity matrix 
Create a 3x3 matrix of your choice and check if it's invertible 

```python
matrix = np.identity(3)
print(matrix)
```

```python
m = np.random.randint(1,101,size = (3,3))
print(m)
```

```python
print(matrix @ m)
```

```python
print(np.linalg.det(matrix))
```

```python
m = m/np.linalg.norm(m)
print(np.linalg.det(m))
```

```python
print(np.linalg.inv(m))
```

Reshape it to (10, 64) — flatten each board into a single row
