```python
import numpy as np

a = np.array([10, 25, 30, 45, 50])
result = np.where(a > 30, 1, 0)
print(result)
```

```python
a = np.array([10, 25, 30, 45, 50])
result = np.where(a > 30, a, 0)
print(result)
```

```python
a = np.array([1,2,3])
b = np.array([4,5,6])
print(np.concatenate([a, b]))
```

```python
a = np.array([[1,2,3],
              [4,5,6]])

b = np.array([[7,8,9],
              [10,11,12]])

print(np.concatenate([a, b], axis=0))
print(np.concatenate([a, b], axis=1))
```

```python
a = np.array([1,2,3])
b = np.array([4,5,6])

print(np.concatenate([a,b]))
print(np.stack([a,b]))
```

```python
a = np.ones((4,5))
b = np.ones((4,5))
c = np.ones((4,5))
print(np.stack([a,b,c]).shape)
```

```python
a = np.array([[1,2],
              [3,4]])

print(np.linalg.det(a))
print(np.linalg.inv(a))
print(np.linalg.norm(a))
```
