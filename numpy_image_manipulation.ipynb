```python
from PIL import Image
import numpy as np
```

```python
img = Image.open("doom.jpeg")
arr = np.array(img)
print(arr.shape)
print(arr.dtype)
```

```python
flipped = arr[:, ::-1,:]
fli_image = Image.fromarray(flipped)
fli_image
```

```python
gray = arr.mean(axis=2)
print(gray.shape)
print(gray.dtype)
```

```python
gray_uint8 = gray.astype(np.uint8)
gray_img = Image.fromarray(gray_uint8)
gray_img
```

```python

sliced = gray[int(gray.shape[0]/4):int(gray.shape[0]/2+gray.shape[0]/4),:]
sliced_image = Image.fromarray(sliced)
sliced_image.show()
```
