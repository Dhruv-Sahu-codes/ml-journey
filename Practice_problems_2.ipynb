```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv')
```

# What was the overall survival rate?

```python
print("Total number of people : ",df['Survived'].count())
print("Number of Survivers : ",df['Survived'].sum())
print("Survival rate : ",df['Survived'].mean())
```

```python
df['Survived'].value_counts().plot(kind = 'bar')
plt.ylabel('Count')
plt.show()
```

# Did gender affect survival? (visualize it)

```python
df.groupby('Sex')['Survived'].mean().plot(kind = 'bar')
plt.ylabel('Count')
```

# Did passenger class affect survival? (visualize it)

```python
df.groupby('Pclass')['Survived'].mean().plot(kind = 'bar')
plt.ylabel('Count')
```

# What was the age distribution of survivors vs non-survivors?

```python
df[df['Survived'] == 1]['Age'].plot(kind='hist',alpha = 0.5, label='Survived')
df[df['Survived'] == 0]['Age'].plot(kind='hist', alpha = 0.5,label='Not Survived')
plt.legend()
```

# Did fare paid correlate with survival?

```python
df[df['Survived'] == 1]['Fare'].plot(kind='hist', alpha=0.5, label='Survived')
df[df['Survived'] == 0]['Fare'].plot(kind='hist', alpha=0.5, label='Not Survived')
plt.legend()
plt.xlabel('Fare')
plt.show()
```
