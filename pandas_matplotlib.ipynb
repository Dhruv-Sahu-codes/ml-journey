```python
import matplotlib.pyplot as plt
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv')
```

```python
ages = df['Age'].dropna()
plt.hist(ages,bins = [0,5,10,15,20,25,30,35,40,45,50,55,60,65,70,75,80])
plt.xlabel('Age')
plt.ylabel('Count')
plt.title('Age Distribution')
plt.show()
```

```python
plt.hist(ages, bins=[0,5,10,15,20,25,30,35,40,45,50,55,60,65,70,75,80])
plt.xticks([0,5,10,15,20,25,30,35,40,45,50,55,60,65,70,75,80])
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
df['Pclass'].value_counts().plot(kind='bar')
plt.xlabel('Passenger Class')
plt.ylabel('Count')
plt.title('Passengers per Class')
plt.show()
```

```python
df.groupby('Sex')['Survived'].sum().plot(kind='bar')
plt.ylabel('Survival Rate')
plt.title('Survival Rate by Gender')
plt.show()
```

```python
plt.scatter(df['Age'], df['Fare'])
plt.xlabel('Age')
plt.ylabel('Fare')
plt.title('Age vs Fare')
plt.show()
```

```python
plt.scatter(df['Pclass'], df['Fare'])
plt.xlabel('Pclass')
plt.ylabel('Fare')
plt.title('Pclass vs Fare')
plt.show()
```

```python
df['Pclass'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.ylabel('')
plt.title('Passenger Class Distribution')
plt.show()
```

```python
df.groupby('Pclass')['Fare'].mean().plot(kind='line')
```

```python
df.groupby('Pclass')['Fare'].mean().plot(kind='line', label='Average Fare')
df.groupby('Pclass')['Age'].mean().plot(kind='line', label='Average Age')
plt.legend()
plt.show()
```
