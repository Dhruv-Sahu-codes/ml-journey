```python
df = pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv')
print(df.shape)
print(df.head())
```

```python
print(df.isnull().sum())
```

```python
print(df['Age'])
print(df[['Age', 'Survived']])
```

```python
print(df.iloc[0])
print(df.loc[0])
```

```python
s = pd.Series([10, 20, 30, 40, 50], index=['a', 'b', 'c', 'd', 'e'])
print(s.iloc[0])   
print(s.loc['a'])  
print(s.loc[0])    
```

```python
print(df.loc[0:3, ['Name', 'Age', 'Survived']])
print(df.iloc[0:3, 0:3])
```

```python
survivors = df[df['Survived'] == 1]
print(survivors.shape)
```

```python
female_survivors = df[(df['Survived'] == 1) & (df['Sex'] == 'female')]
print(female_survivors.shape)
```

```python
print(df.groupby('Sex')['Survived'].mean())
```

```python
print(df.groupby('Pclass')['Survived'].mean())
```

```python
print(df['Sex'].value_counts())
print(df['Pclass'].value_counts())
```

```python
print(df.groupby('Sex')['Survived'].value_counts())
print(df.groupby('Sex')['Survived'].sum())
```
